1. *Runtime → Change runtime type → **T4 GPU*** → Save
2. Run the 3 cells. The last one prints the **demo link** (about 10 min). Full instructions are in `COLAB_STEPS.md`.

```
# 1) Credentials (hidden input, never stored in the notebook)
import os, getpass
user = input("GitHub username: ").strip()
repo = input("Repository name [prompt-injection-guard]: ").strip() or "prompt-injection-guard"
token = getpass.getpass("GitHub token (Enter if the repo is public): ")
os.environ["REPO_URL"] = f"https://{user}:{token}@github.com/{user}/{repo}.git" if token else f"https://github.com/{user}/{repo}.git"
os.environ["KAGGLE_USERNAME"] = input("Kaggle username (Enter to skip): ").strip()
os.environ["KAGGLE_KEY"] = getpass.getpass("Kaggle key: ") if os.environ["KAGGLE_USERNAME"] else ""
RETRAIN = input("Retrain everything on the GPU first? (y/N): ").strip().lower() == "y" 
```

```
# 2) Download the project
!rm -rf /content/proj && git clone -q "$REPO_URL" /content/proj && git -C /content/proj remote set-url origin "https://github.com/$(basename $(dirname "$REPO_URL"))/$(basename "$REPO_URL")" && ls /content/proj
```

```
# 3) Install everything + start the demo (prints the public link at the end)
FLAG = "--retrain" if RETRAIN else ""
!cd /content/proj && bash cloud/start.sh $FLAG
```
